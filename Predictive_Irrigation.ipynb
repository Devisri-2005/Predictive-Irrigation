```python
import pandas as pd
import numpy as np

path = "/content/drive/MyDrive/Predictive_Irrigation/DATA/processed/environmental_soil_dataset.csv"

# Load saved dataset
merged_data = pd.read_csv(path)

# Convert Date and sort
merged_data["Date"] = pd.to_datetime(merged_data["Date"])
merged_data = merged_data.sort_values("Date").reset_index(drop=True)

# 1. Future soil-moisture target
merged_data["Target_SM_next"] = (
    merged_data["SMAP_soil moisture"].shift(-1)
)

ml_data = merged_data.dropna(
    subset=["Target_SM_next"]
).copy()

# 2. Soil-moisture lag features
ml_data["SM_lag_1"] = ml_data["SMAP_soil moisture"].shift(1)
ml_data["SM_lag_2"] = ml_data["SMAP_soil moisture"].shift(2)
ml_data["SM_lag_3"] = ml_data["SMAP_soil moisture"].shift(3)

ml_data = ml_data.dropna(
    subset=["SM_lag_1", "SM_lag_2", "SM_lag_3"]
).copy()

# 3. Rainfall-history features
ml_data["Rainfall_lag_1"] = (
    ml_data["Rainfall_mm_day"].shift(1)
)

ml_data["Rainfall_3obs_sum"] = (
    ml_data["Rainfall_mm_day"].rolling(3).sum()
)

ml_data["Rainfall_7obs_sum"] = (
    ml_data["Rainfall_mm_day"].rolling(7).sum()
)

ml_data = ml_data.dropna(
    subset=[
        "Rainfall_lag_1",
        "Rainfall_3obs_sum",
        "Rainfall_7obs_sum"
    ]
).copy()

# 4. Seasonal features
ml_data["Month"] = ml_data["Date"].dt.month
ml_data["Day_of_Year"] = ml_data["Date"].dt.dayofyear

ml_data["Month_sin"] = np.sin(
    2 * np.pi * ml_data["Month"] / 12
)

ml_data["Month_cos"] = np.cos(
    2 * np.pi * ml_data["Month"] / 12
)

ml_data = ml_data.reset_index(drop=True)

print("Feature-engineered ML dataset recreated successfully!")
print("Shape:", ml_data.shape)
print("Missing values:", ml_data.isnull().sum().sum())
```

```python
feature_columns = [
    "SMAP_soil moisture",
    "Temperature_C",
    "Humidity_Percent",
    "Rainfall_mm_day",
    "Solar_Radiation_MJ_m2_day",
    "SM_lag_1",
    "SM_lag_2",
    "SM_lag_3",
    "Rainfall_lag_1",
    "Rainfall_3obs_sum",
    "Rainfall_7obs_sum",
    "Month_sin",
    "Month_cos"
]

# Input features
X = ml_data[feature_columns].copy()

# Prediction target
y = ml_data["Target_SM_next"].copy()

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeatures:")
for i, feature in enumerate(X.columns, start=1):
    print(f"{i}. {feature}")

print("\nTarget:", y.name)

print("\nMissing values in X:", X.isnull().sum().sum())
print("Missing values in y:", y.isnull().sum())
```

```python
# Number of observations
n = len(X)

# 80% for training
train_size = int(n * 0.80)

# Chronological split
X_train = X.iloc[:train_size].copy()
X_test = X.iloc[train_size:].copy()

y_train = y.iloc[:train_size].copy()
y_test = y.iloc[train_size:].copy()

print("Chronological train/test split completed!")

print("\nTraining data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

print("\nTraining period:")
print(
    ml_data["Date"].iloc[0],
    "to",
    ml_data["Date"].iloc[train_size - 1]
)

print("\nTesting period:")
print(
    ml_data["Date"].iloc[train_size],
    "to",
    ml_data["Date"].iloc[-1]
)
```

```python
from sklearn.linear_model import LinearRegression

# Create the Linear Regression model
lr_model = LinearRegression()

# Train the model
lr_model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Predict on unseen test data
y_pred_lr = lr_model.predict(X_test)

# Calculate evaluation metrics
mae_lr = mean_absolute_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_test, y_pred_lr))
r2_lr = r2_score(y_test, y_pred_lr)

print("Linear Regression Results")
print("-------------------------")
print(f"MAE  : {mae_lr:.6f}")
print(f"RMSE : {rmse_lr:.6f}")
print(f"R²   : {r2_lr:.6f}")
```

```python
from sklearn.ensemble import RandomForestRegressor

# Create Random Forest model
rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

# Train the model
rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
# Predict on unseen test data
y_pred_rf = rf_model.predict(X_test)

# Calculate evaluation metrics
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest Results")
print("---------------------")
print(f"MAE  : {mae_rf:.6f}")
print(f"RMSE : {rmse_rf:.6f}")
print(f"R²   : {r2_rf:.6f}")
```

```python
from sklearn.ensemble import GradientBoostingRegressor

# Create Gradient Boosting model
gb_model = GradientBoostingRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

# Train the model
gb_model.fit(X_train, y_train)

print("Gradient Boosting model trained successfully!")
```

```python
# Predict on unseen test data
y_pred_gb = gb_model.predict(X_test)

# Calculate evaluation metrics
mae_gb = mean_absolute_error(y_test, y_pred_gb)
rmse_gb = np.sqrt(mean_squared_error(y_test, y_pred_gb))
r2_gb = r2_score(y_test, y_pred_gb)

print("Gradient Boosting Results")
print("------------------------")
print(f"MAE  : {mae_gb:.6f}")
print(f"RMSE : {rmse_gb:.6f}")
print(f"R²   : {r2_gb:.6f}")
```

```python
# Persistence baseline:
# Predict the next soil-moisture observation as the current soil moisture

y_pred_persistence = X_test["SMAP_soil moisture"].values

# Calculate evaluation metrics
mae_persistence = mean_absolute_error(y_test, y_pred_persistence)
rmse_persistence = np.sqrt(mean_squared_error(y_test, y_pred_persistence))
r2_persistence = r2_score(y_test, y_pred_persistence)

print("Persistence Baseline Results")
print("----------------------------")
print(f"MAE  : {mae_persistence:.6f}")
print(f"RMSE : {rmse_persistence:.6f}")
print(f"R²   : {r2_persistence:.6f}")
```

```python
# Percentage improvement of Gradient Boosting over Persistence Baseline

mae_improvement = (
    (mae_persistence - mae_gb) / mae_persistence
) * 100

rmse_improvement = (
    (rmse_persistence - rmse_gb) / rmse_persistence
) * 100

print("Gradient Boosting Improvement over Persistence Baseline")
print("-------------------------------------------------------")
print(f"MAE improvement  : {mae_improvement:.2f}%")
print(f"RMSE improvement : {rmse_improvement:.2f}%")
print(f"R² improvement   : {r2_gb - r2_persistence:.6f}")
```

```python
# Evaluate AI irrigation decisions against the actual next available SMAP condition

test_results_improved["Actual_Next_Threshold"] = (
    test_results_improved["Target_SM_next"] <= irrigation_threshold
).astype(int)

# Confusion matrix components
TP = (
    (test_results_improved["AI_Irrigation"] == 1) &
    (test_results_improved["Actual_Next_Threshold"] == 1)
).sum()

TN = (
    (test_results_improved["AI_Irrigation"] == 0) &
    (test_results_improved["Actual_Next_Threshold"] == 0)
).sum()

FP = (
    (test_results_improved["AI_Irrigation"] == 1) &
    (test_results_improved["Actual_Next_Threshold"] == 0)
).sum()

FN = (
    (test_results_improved["AI_Irrigation"] == 0) &
    (test_results_improved["Actual_Next_Threshold"] == 1)
).sum()

accuracy = (TP + TN) / len(test_results_improved)

precision = TP / (TP + FP) if (TP + FP) > 0 else 0
recall = TP / (TP + FN) if (TP + FN) > 0 else 0
specificity = TN / (TN + FP) if (TN + FP) > 0 else 0

print("AI Irrigation Decision Evaluation")
print("==================================")

print(f"True Positives  (TP): {TP}")
print(f"True Negatives  (TN): {TN}")
print(f"False Positives (FP): {FP}")
print(f"False Negatives (FN): {FN}")

print("\nDecision Agreement Metrics")
print(f"Accuracy    : {accuracy * 100:.2f}%")
print(f"Precision   : {precision * 100:.2f}%")
print(f"Recall      : {recall * 100:.2f}%")
print(f"Specificity : {specificity * 100:.2f}%")
```
